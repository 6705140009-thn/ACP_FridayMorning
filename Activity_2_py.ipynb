## Part 0 — Warm-up: from a function to an object

Last week you organised code with **functions**. This week we bundle data **and** the functions that work on it into a single **object**. First, run this reminder of the old way:

```python
# The 192-101 way: data and function are separate
balance = 100

def deposit(current, amount):
    return current + amount

balance = deposit(balance, 50)
print(balance)   # 150
```

> Nothing links `balance` and `deposit` together — you must pass the balance in every time. A **class** fixes this by keeping them in one place. That is what we build today.

## Part 1 — Define a class and make an object

A **class** is a blueprint for a kind of object. Use the `class` keyword and a `CapWords` or  `Pascal Case` name. Run this tiny class:

```python
class Dog:
    """A dog with a name."""

    def __init__(self, name):   # runs automatically when you make a Dog
        self.name = name        # store the name on THIS object
d1 = Dog('Rex')                 # d1 is an object (an instance) of Dog
print(d1.name)                  # Rex
```

Key words to notice:
- **class** `Dog` — the blueprint.
- **object / instance** `d1` — a real dog built from the blueprint.
- **attribute** `name` — data stored on the object, read with `d1.name`.
- **`__init__`** — the special method that runs automatically to set up a new object.
- **`self`** — the object being built or used (here, `d1`).

**Your turn.** Make a class called `Cat` with a `name` attribute (copy the pattern above). Then create a cat called `'Milo'` and print its name.

```python
# Your code here:
class Cat:
    def __init__(self, name):
        self.name = name

c1 = Cat('Milo')
print(c1.name)
```

## Part 2 — Objects can hold several attributes

An object usually stores more than one piece of data. Pass several values into `__init__` and store each with `self.`. This is the **Basic Class and Object** example from your workbook *(Python 100 Programs, SET-2 #10)*.

```python
class Rectangle:
    """A rectangle with a length and a width."""

    def __init__(self, length, width):
        self.length = length    # attribute 1
        self.width = width      # attribute 2

r = Rectangle(4, 5)
print(r.length)   # 4
print(r.width)    # 5
```

**Your turn.** Make a class `Student` with **two** attributes set in `__init__`: `name` and `student_id`. Create a student `'Mai'` with id `6501`, then print both attributes.

```python
# Your code here:
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id

s = Student('Mai', 6501)
print(s.name)
print(s.student_id)
```

## Part 3 — Methods: behaviour that uses the object's data

A **method** is a function defined *inside* a class. It always takes `self` first, so it can use the object's own attributes. Here we add an `area()` method to `Rectangle` — completing the workbook example:

```python
class Rectangle:
    """A rectangle with a length, a width, and an area."""

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):                     # a method (behaviour)
        return self.length * self.width

r = Rectangle(4, 5)
print(r.area())   # 20  <- called ON the object with r.area()
```

Notice `r.area()` is called **on the object**. Inside the method, `self` *is* `r`, so `self.length` is `4` and `self.width` is `5`.

**Your turn.** Add a method `perimeter(self)` to the `Rectangle` class that returns `2 * (length + width)`. Then create a rectangle and print its perimeter.

*Tip:* copy the class above into the cell, then add your new method next to `area`.

```python
# Your code here:
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

r = Rectangle(4, 5)
print(r.perimeter())  # 18
```

## Part 4 — Make your object readable with `__str__`

By default, printing an object shows something unhelpful like `<__main__.Rectangle object at 0x...>`. Run this to see the problem:

```python
r = Rectangle(4, 5)
print(r)     # unhelpful memory address
```

Add a special method `__str__` that **returns** a readable string. Python calls it automatically whenever you `print` the object:

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def __str__(self):                  # decides how the object looks as text
        return f'Rectangle {self.length}x{self.width} = {self.area()}'

r = Rectangle(4, 5)
print(r)     # Rectangle 4x5 = 20
```

> Remember: `__str__` must **return** a string (not `print` it). Then `print(r)` and `str(r)` both use it.

**Your turn.** Add a `__str__` to your `Student` class from Part 2 so that `print(student)` shows: `Student: Mai (6501)`.

```python
# Your code here:
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id

    def __str__(self):
        return f'Student: {self.name} ({self.student_id})'

student = Student('Mai', 6501)
print(student)  # Student: Mai (6501)
```

## Part 5 — Guided build: a `BankAccount` class

Now put it all together — attributes, methods, and `__str__` — in one class. This mirrors the **Counter** class in your workbook *(Python 100 Programs, SET-10 #8)*, where methods change an attribute.

Fill in the `# TODO` lines. The first method is done for you as an example.

```python
class BankAccount:
    """A simple bank account."""

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        self.balance = self.balance - amount

    def __str__(self):
        return f'{self.owner}: ${self.balance}'

acct = BankAccount('Mai', 100)
acct.deposit(50)
acct.withdraw(30)
print(acct.balance)      # 120
print(acct)              # Mai: $120
```

## Part 6 — Build it yourself: a `Student` with grades

From this short spec, write a `Student` class **from scratch** (this is the core Week 2 skill). Type it yourself — **AI-use Level 1**.

**Specification**
1. **Attributes** (set in `__init__`): `name`, `student_id`, and an empty list `grades`.
2. **Method** `add_grade(self, g)` — append `g` to `self.grades`.
3. **Method** `average(self)` — return the mean of `self.grades` (return `0` if the list is empty).
4. **`__str__`** — return e.g. `Mai (6501): avg 3.50`.
5. **Exercise it** — create 2 students, add a few grades to each, and print each one.

```python
class Student:
    """A student with a name, id, and grades."""

    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.grades = []

    def add_grade(self, g):
        self.grades.append(g)

    def average(self):
        if len(self.grades) == 0:
            return 0
        return sum(self.grades) / len(self.grades)

    def __str__(self):
        return f'{self.name} ({self.student_id}): avg {self.average():.2f}'

# Create 2 students, add grades, and print
s1 = Student('Mai', 6501)
s1.add_grade(3.5)
s1.add_grade(4.0)
s1.add_grade(3.0)

s2 = Student('Alex', 6502)
s2.add_grade(3.8)
s2.add_grade(3.9)

print(s1)  # Mai (6501): avg 3.50
print(s2)  # Alex (6502): avg 3.85
```
